Definicion de funciones matematicas

```python
import numpy as np

# Definición de la función de activación sigmoide
def sigmoide(x):
    return 1 / (1 + np.exp(-x))

# Definición de la derivada de la sigmoide 
def derivada_sigmoide(salida_sigmoide):
    # Matemáticamente es: sigmoide(x) * (1 - sigmoide(x))
    return salida_sigmoide * (1 - salida_sigmoide)
```

Configuracion de entradas

```python
# Entradas del sistema
entradas = np.array([0.35, 0.9])

# Matriz de pesos hacia la capa oculta. 
# Fila 1: pesos de las entradas hacia la neurona A (0.1 y 0.8)
# Fila 2: pesos de las entradas hacia la neurona B (0.4 y 0.6)
pesos_entrada_oculta = np.array([[0.1, 0.8],  
                                 [0.4, 0.6]]) 

# Vector de pesos desde las neuronas ocultas A y B hacia la neurona de salida C
pesos_oculta_salida = np.array([0.3, 0.9])    

# parametros de entrenamiento
objetivo = 0.5
tasa_aprendizaje = 0.5
```

Calculo de salida (objetico 1)

```python
print("--- i) CALCULO DE LA SALIDA ACTUAL ---")

# De las entradas hacia las Neuronas ocultas A y B
entrada_neta_oculta = np.dot(pesos_entrada_oculta, entradas)
salida_oculta = sigmoide(entrada_neta_oculta) # devuelve  [Salida_A, Salida_B]

# De las Neuronas A y B hacia la Neurona de salida C
entrada_neta_salida = np.dot(pesos_oculta_salida, salida_oculta)
salida_final = sigmoide(entrada_neta_salida)

print(f"Activación Neurona A: {salida_oculta[0]:.5f}")
print(f"Activación Neurona B: {salida_oculta[1]:.5f}")
print(f"Salida total inicial (Neurona C): {salida_final:.5f}")
```

Entrenamiento de reversa (objeivo 2)

```python
print("\n--- ii) ENTRENAMIENTO DE REVERSA ---")

# 1. Calcular el error en la capa de salida 
error_salida = salida_final - objetivo
gradiente_salida = error_salida * derivada_sigmoide(salida_final)

# 2. Propagar el error hacia la capa oculta
error_oculta = gradiente_salida * pesos_oculta_salida
gradiente_oculta = error_oculta * derivada_sigmoide(salida_oculta)

# 3. Actualización de pesos 
pesos_oculta_salida_nuevos = pesos_oculta_salida - (tasa_aprendizaje * gradiente_salida * salida_oculta)
pesos_entrada_oculta_nuevos = pesos_entrada_oculta - (tasa_aprendizaje * np.outer(gradiente_oculta, entradas))

print(f"Nuevos pesos hacia la Neurona C (wAC, wBC): \n{pesos_oculta_salida_nuevos}")
print(f"Nuevos pesos hacia las Neuronas A y B: \n{pesos_entrada_oculta_nuevos}")
```

Recalcular salida total (objetivo 3)

```python
print("\n--- iii) RECALCULAR LA SALIDA TOTAL ---")

# Forward pass utilizando exclusivamente las matrices de pesos actualizadas
entrada_neta_oculta_nueva = np.dot(pesos_entrada_oculta_nuevos, entradas)
salida_oculta_nueva = sigmoide(entrada_neta_oculta_nueva)

entrada_neta_salida_nueva = np.dot(pesos_oculta_salida_nuevos, salida_oculta_nueva)
salida_final_nueva = sigmoide(entrada_neta_salida_nueva)

print(f"Nueva salida total tras 1 iteración (Neurona C): {salida_final_nueva:.5f}")
```

Segunda epoca


```python
print("\n--- iv) SEGUNDA ÉPOCA DE ENTRENAMIENTO ---")

# 1. Calcular el nuevo gradiente en la capa de salida con los valores de la epoca 1
error_salida_2 = salida_final_nueva - objetivo
gradiente_salida_2 = error_salida_2 * derivada_sigmoide(salida_final_nueva)

# 2. Propagar el error hacia la capa oculta
error_oculta_2 = gradiente_salida_2 * pesos_oculta_salida_nuevos
gradiente_oculta_2 = error_oculta_2 * derivada_sigmoide(salida_oculta_nueva)

# 3. Segunda actualizacion de pesos
pesos_oculta_salida_epoca2 = pesos_oculta_salida_nuevos - (tasa_aprendizaje * gradiente_salida_2 * salida_oculta_nueva)
pesos_entrada_oculta_epoca2 = pesos_entrada_oculta_nuevos - (tasa_aprendizaje * np.outer(gradiente_oculta_2, entradas))

print(f"Pesos actualizados hacia C (epoca 2): \n{pesos_oculta_salida_epoca2}")
print(f"Pesos actualizados hacia A y B (epoca 2): \n{pesos_entrada_oculta_epoca2}")

# 4. Tercer calculo de salida (Forward pass de la Época 2)
entrada_neta_oculta_epoca2 = np.dot(pesos_entrada_oculta_epoca2, entradas)
salida_oculta_epoca2 = sigmoide(entrada_neta_oculta_epoca2)

entrada_neta_salida_epoca2 = np.dot(pesos_oculta_salida_epoca2, salida_oculta_epoca2)
salida_final_epoca2 = sigmoide(entrada_neta_salida_epoca2)

print(f"\nSalida total tras la 2da epoca (Neurona C): {salida_final_epoca2:.5f}")
```

Enrenamiento hasta el objetivo

```python
print("\n--- v) ENTRENAMIENTO AUTOMATIZADO HASTA CONVERGENCIA ---")

# Reiniciamos las matrices de pesos a sus valores iniciales del diagrama
pesos_entrada_oculta_bucle = np.array([[0.1, 0.8], 
                                       [0.4, 0.6]]) 
pesos_oculta_salida_bucle = np.array([0.3, 0.9])

epocas_totales = 0
tolerancia_error = 0.0001
salida_bucle = 1.0 # Inicializamos con un valor temporal

# El bucle se ejecuta mientras el error absoluto sea mayor a la tolerancia (0.0001)
while abs(salida_bucle - objetivo) > tolerancia_error:
    
    # 1. Forward Propagacion
    entrada_oculta = np.dot(pesos_entrada_oculta_bucle, entradas)
    sal_oculta = sigmoide(entrada_oculta)
    
    entrada_salida = np.dot(pesos_oculta_salida_bucle, sal_oculta)
    salida_bucle = sigmoide(entrada_salida)
    
    # Verificamos si ya llegamos a la meta para detener el ciclo
    if abs(salida_bucle - objetivo) <= tolerancia_error:
        break
        
    # 2. propagacion atras
    error_salida_bucle = salida_bucle - objetivo
    grad_salida_bucle = error_salida_bucle * derivada_sigmoide(salida_bucle)
    
    error_oculta_bucle = grad_salida_bucle * pesos_oculta_salida_bucle
    grad_oculta_bucle = error_oculta_bucle * derivada_sigmoide(sal_oculta)
    
    # 3. Actualizacion de pesos
    pesos_oculta_salida_bucle -= (tasa_aprendizaje * grad_salida_bucle * sal_oculta)
    pesos_entrada_oculta_bucle -= (tasa_aprendizaje * np.outer(grad_oculta_bucle, entradas))
    
    epocas_totales += 1

print(f"¡Convergencia matemática alcanzada!")
print(f"epocas totales requeridas de entrenamiento: {epocas_totales}")
print(f"Salida final lograda: {salida_bucle:.5f}")
```
